**AI vs ML vs DL**

**AI**-Describes how computer and technology mimices human brian, smart applications that can perform a task without human intervention

**ML**- Focuses on creating algorithms and stastical models to let computers learn and make predictions without explicit programming.

**DL**-Specialized ML algorithms that mimices human brian.

**Types of ML**

**1)Supervised ML**- The ML style where dependent variable y is present.
    y=f(x)
-   In regression, y is continuous and in classification y is discrete.

**2)Unsupervised ML** - There is no target y and no historical data.
     - can you find grouping in the data, for example customer segmentation.

**3)Combination of supervised and unsupervised**- Includes data with both labelled and unlabelled data.Netflix and amazon prime.

**4) Reinforcement Learning** - Concerned with how intelligent agent take action in an environment to maximize the reward.

**Train Test and Validation**

**Train** - To train the model

**Validate** - Hyperparameter tuning of model.

**Test** - Test the model accuracy.

**Underfitting, Overfitting and Bias variance**

**Model Training** - Given the data estimate the prediction functon by minimizing the error.

**Overfitting** - Model performs well on train data but worst on test data.
**Underfitting** -  Model performs poor on both on training and testing data.

-   In generalized model, the accuracy is good on both at training and testing.
-   Training error is known as bias.
-   Testing error is also known as variance.

**Misiing values and its Treatment**

**1)Missing completely at random(RCAM) ** - Missing data is independent of observed and unobserved data/missing values.

**2)Missing at Random (MAR)** - missing data depends on observed data but not on missing data itself.

**3)Missing data not at random (MNAR)** - Missing values depends on the value of missing data itself. Eg:- income and job satisfaction of employees in company.

-   **Treatment**
**WHY** - For analysis and ML prediction , its important to deal with missing values.
